## (2.1.1) 2D transformations
### Rotation + translation (aks Similarity)

```python
%matplotlib widget
fig, ax = plt.subplots(figsize = (8,4))
x = np.array([[0,1,1,0,0],[0,0,1,1,0]])
xbar = np.r_[x, np.ones((1,5))]
tta = np.radians(10)
t = np.array([1, 0.5])
R = np.array([[np.cos(tta), -np.sin(tta)],[np.sin(tta), np.cos(tta)]])
s = .2
T = np.c_[s * R, t]
xt = np.matmul(T, xbar)

ax.plot(x[0], x[1], 'o-')
ax.plot(xt[0], xt[1], 'ro-')
ax.grid()
ax.axis('equal')
```
