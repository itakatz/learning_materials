```python
import matplotlib.pyplot as plt
import numpy as np
from numpy.linalg import norm
import scipy.spatial as spatial
from scipy.spatial.transform import Rotation as R
```

```python
def generate_random_convex_polygon(n):
    """
    Generate a random convex polygon with n vertices.
    
    Parameters:
    -----------
    n : int
        Number of vertices for the polygon
    
    Returns:
    --------
    numpy.ndarray
        Array of (x,y) coordinates forming a convex polygon
    """
    # Generate random points in 2D space
    points = 2 * np.random.randn(n, 2) - 1.
    
    # Compute the convex hull of these points
    # This ensures the polygon is convex
    hull = spatial.ConvexHull(points)
    
    # Reorder points to form the convex polygon
    convex_polygon = points[hull.vertices]
    
    return convex_polygon
```
